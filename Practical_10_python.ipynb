```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df['species'] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

print(df.head())

print(df.info())

print(df.describe())
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='sepal length (cm)',
    y='petal length (cm)',
    hue='species',
    s=80
)

plt.title('Scatter Plot of Iris Dataset')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x='petal length (cm)',
    hue='species',
    fill=True,
    alpha=0.3
)

plt.title('Density Plot of Petal Length')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='species',
    y='petal length (cm)'
)

plt.title('Violin Plot of Iris Species')
plt.show()
```

```python
plt.figure(figsize=(9, 6))

correlation = df.select_dtypes(
    include='number'
).corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Iris Dataset')
plt.show()
```

```python
sns.jointplot(
    data=df,
    x='sepal length (cm)',
    y='petal length (cm)',
    hue='species',
    height=7
)

plt.show()
```

```python
sns.pairplot(
    df,
    hue='species',
    diag_kind='hist'
)

plt.show()
```
